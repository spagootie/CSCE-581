#Imports and configuration

```python
from __future__ import annotations

import json
from pathlib import Path
from typing import Dict, List

import numpy as np
import pandas as pd
import requests
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, f1_score, roc_auc_score
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import Pipeline

from lime.lime_text import LimeTextExplainer

REPO_API = "https://api.github.com/repos/ai4society/election-dataset-us/contents/data-us?ref=main"
OUT_CSV = Path("us-allstates-voter-faqs.csv")
OUT_METRICS = Path("metrics.json")
OUT_LIME_DIR = Path("lime_explanations")

pd.set_option("display.max_colwidth", 120)
```

#Helper functions

```python
def fetch_repo_listing(session: requests.Session) -> List[dict]:
    resp = session.get(REPO_API, timeout=60)
    resp.raise_for_status()
    items = resp.json()
    if not isinstance(items, list):
        raise ValueError("Unexpected GitHub API response; expected a list of files.")
    return items


def infer_label(source_url: str) -> int:
    """Map source URL to label.

    0 = primary (official / state source)
    1 = secondary (Vote411 / LWV or other non-official source)
    """
    src = (source_url or "").lower()
    if "vote411.org" in src or "lwv" in src:
        return 1
    return 0


def build_text(question: str, answer: str) -> str:
    q = " ".join(str(question).split())
    a = " ".join(str(answer).split())
    return f"Question: {q} Answer: {a}"


def load_samples() -> pd.DataFrame:
    session = requests.Session()
    session.headers.update({"User-Agent": "Mozilla/5.0 (compatible; model-builder/1.0)"})
    listing = fetch_repo_listing(session)

    rows: List[Dict[str, object]] = []
    for item in listing:
        name = item.get("name", "")
        if not name.endswith("_qa.json"):
            continue
        download_url = item.get("download_url")
        if not download_url:
            continue

        r = session.get(download_url, timeout=60)
        r.raise_for_status()
        data = r.json()

        state = str(data.get("state", name[:2].upper()))
        questions = data.get("questions", [])
        if not isinstance(questions, list):
            continue

        for qa in questions:
            if not isinstance(qa, dict):
                continue
            rows.append({
                "State": state,
                "text": build_text(qa.get("q", ""), qa.get("a", "")),
                "label": infer_label(str(qa.get("s", ""))),
            })

    if not rows:
        raise RuntimeError("No rows were loaded. Check network access or repo status.")

    return pd.DataFrame(rows, columns=["State", "text", "label"])


def fit_and_score(name: str, pipeline: Pipeline, X_train, X_test, y_train, y_test) -> dict:
    pipeline.fit(X_train, y_train)
    y_pred = pipeline.predict(X_test)

    if hasattr(pipeline, "predict_proba"):
        y_score = pipeline.predict_proba(X_test)[:, 1]
    else:
        y_score = pipeline.decision_function(X_test)

    return {
        "model": name,
        "f1": float(f1_score(y_test, y_pred)),
        "roc_auc": float(roc_auc_score(y_test, y_score)),
    }


def build_models():
    vectorizer = TfidfVectorizer(
        stop_words="english",
        ngram_range=(1, 2),
        min_df=2,
        max_df=0.95,
    )

    logistic = Pipeline(steps=[
        ("tfidf", vectorizer),
        ("clf", LogisticRegression(max_iter=3000, class_weight="balanced", solver="liblinear")),
    ])

    nb = Pipeline(steps=[
        ("tfidf", vectorizer),
        ("clf", MultinomialNB(alpha=0.5)),
    ])

    return logistic, nb


def save_lime_explanations(model: Pipeline, X_test: pd.Series, y_test: pd.Series) -> None:
    OUT_LIME_DIR.mkdir(parents=True, exist_ok=True)
    class_names = ["primary", "secondary"]
    explainer = LimeTextExplainer(class_names=class_names, random_state=42)

    preds = model.predict(X_test)

    for target_label, label_name in [(0, "primary"), (1, "secondary")]:
        mask = (y_test == target_label) & (preds == target_label)
        if mask.any():
            idx = X_test[mask].index[0]
        else:
            idx = X_test[y_test == target_label].index[0]

        text = X_test.loc[idx]
        pred_label = int(model.predict([text])[0])
        exp = explainer.explain_instance(
            text_instance=text,
            classifier_fn=model.predict_proba,
            num_features=12,
            labels=[1],
        )

        html_path = OUT_LIME_DIR / f"lime_{label_name}_example.html"
        exp.save_to_file(str(html_path))

        txt_path = OUT_LIME_DIR / f"lime_{label_name}_example.txt"
        with txt_path.open("w", encoding="utf-8") as f:
            f.write(f"Actual label: {label_name}\n")
            f.write(f"Predicted: {class_names[pred_label]}\n\n")
          #  f.write("Top features for primary (class 0):\n")
          #  for feat, weight in exp.as_list(label=0):
          #      f.write(f"  {feat}: {weight:.4f}\n")
            f.write("\nTop features for secondary (class 1):\n")
            for feat, weight in exp.as_list(label=1):
                f.write(f"  {feat}: {weight:.4f}\n")
```

#Load and combine all state files

```python
df = load_samples()
df.head()
```

#Save the combined CSV and inspect the class balance

```python
df.to_csv(OUT_CSV, index=False)
print(f"Saved combined dataset to {OUT_CSV.resolve()}")
print("Rows:", len(df))
print("Label counts:", df["label"].value_counts().sort_index())
```

#Train/test split

```python
X = df["text"]
y = df["label"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

print("Train size:", len(X_train))
print("Test size:", len(X_test))
```

#Train two models and report F1 / ROC AUC

```python
logistic, nb = build_models()

results = []
for name, model in [("LogisticRegression", logistic), ("MultinomialNB", nb)]:
    metrics = fit_and_score(name, model, X_train, X_test, y_train, y_test)
    results.append(metrics)
    print(f"\n{name}")
    print(f"  F1     : {metrics['f1']:.4f}")
    print(f"  ROC AUC: {metrics['roc_auc']:.4f}")
    print(classification_report(y_test, model.predict(X_test), target_names=["primary", "secondary"]))

results_df = pd.DataFrame(results)
results_df
```

#Generate LIME explanations

```python
save_lime_explanations(logistic, X_test, y_test)
print(f"Saved LIME artifacts to {OUT_LIME_DIR.resolve()}")
```

#Save metrics summary

```python
summary = {
    "dataset_rows": int(len(df)),
    "label_counts": df["label"].value_counts().sort_index().to_dict(),
    "train_size": int(len(X_train)),
    "test_size": int(len(X_test)),
    "models": results,
}

with OUT_METRICS.open("w", encoding="utf-8") as f:
    json.dump(summary, f, indent=2)

summary

from IPython.display import display, HTML
display(HTML(open("lime_explanations/lime_primary_example.html").read()))
display(HTML(open("lime_explanations/lime_secondary_example.html").read()))
```

```python
from matplotlib.backends.backend_pdf import PdfPages
import matplotlib.pyplot as plt
from IPython.display import display

# --- pick the best trained model ---
model_map = {
    "LogisticRegression": logistic,
    "MultinomialNB": nb,
}

best_row = results_df.sort_values(["f1", "roc_auc"], ascending=False).iloc[0]
best_model_name = best_row["model"]
best_model = model_map[best_model_name]

class_names = ["primary", "secondary"]
explainer = LimeTextExplainer(class_names=class_names, random_state=42)

def pick_example_for_state_and_label(dataframe, state, label, seed):
    subset = dataframe[(dataframe["State"] == state) & (dataframe["label"] == label)]
    if not subset.empty:
        return subset.sample(n=1, random_state=seed).iloc[0], ""

    fallback = dataframe[(dataframe["State"] != state) & (dataframe["label"] == label)]
    if fallback.empty:
        raise RuntimeError(f"No fallback example found for state={state}, label={label}")

    row = fallback.sample(n=1, random_state=seed).iloc[0]
    note = f"Missing label {label} for {state}; used {row['State']} instead."
    return row, note

def format_top3(exp, label):
    feats = exp.as_list(label=label)[:3]
    return "; ".join([f"{w} ({v:.4f})" for w, v in feats])

# --- build 100-row sample output ---
sample_rows = []
states = sorted(df["State"].dropna().unique())

pdf_name = f"us-allstates-voter-faqs-classifier-{best_model_name.lower()}-sampleoutput-dump.pdf"
csv_name = f"us-allstates-voter-faqs-classifier-{best_model_name.lower()}-sampleoutput.csv"

with PdfPages(pdf_name) as pdf:
    for i, state in enumerate(states):
        for label in [0, 1]:
            row, note = pick_example_for_state_and_label(df, state, label, seed=1000 + i * 10 + label)

            text = row["text"]
            true_label = int(row["label"])

            probs = best_model.predict_proba([text])[0]
            pred_label = int(np.argmax(probs))
            pred_class = class_names[pred_label]
            true_class = class_names[true_label]
            correct = pred_label == true_label

            exp = explainer.explain_instance(
                text_instance=text,
                classifier_fn=best_model.predict_proba,
                num_features=10,
                labels=[0, 1]
            )

            top3 = format_top3(exp, pred_label)

            sample_rows.append({
                "Text (Q/A)": text,
                "state": state,
                "predicted class": pred_class,
                "true class": true_class,
                "correct": correct,
                "Top three features/words and their weights": top3,
                "note": note
            })

            fig = exp.as_pyplot_figure(label=pred_label)
            fig.set_size_inches(11, 5)
            fig.suptitle(
                f"State: {state} | True: {true_class} | Predicted: {pred_class} | Correct: {correct}",
                fontsize=12
            )
            pdf.savefig(fig, bbox_inches="tight")
            plt.close(fig)

sample_df = pd.DataFrame(sample_rows)

# Save the CSV
sample_df.to_csv(csv_name, index=False)

print(f"Best model: {best_model_name}")
print(f"Saved sample CSV: {csv_name}")
print(f"Saved explainer PDF: {pdf_name}")

display(sample_df.head(10))
```
